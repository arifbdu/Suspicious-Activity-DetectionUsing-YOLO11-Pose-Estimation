```python
# =============================================================
#                 YOLOv11 Evaluation Figures
# =============================================================

from ultralytics import YOLO
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import cv2
import os

# -------------------------
# Load trained model
# -------------------------
model = YOLO("yolov11n.pt")   # replace with your trained weights e.g., "runs/train/exp/weights/best.pt"

# -------------------------
# Run evaluation
# -------------------------
results = model.val()

# Create folder
os.makedirs("figures", exist_ok=True)

# =============================================================
# FIGURE 1: Training Loss Curve
# =============================================================
if os.path.exists("runs/train/exp/results.csv"):
    import pandas as pd

    df = pd.read_csv("runs/train/exp/results.csv")  # automatically generated by Ultralytics

    plt.figure(figsize=(8,5))
    plt.plot(df["train/box_loss"], label="Box Loss")
    plt.plot(df["train/cls_loss"], label="Class Loss")
    plt.plot(df["train/dfl_loss"], label="DFL Loss")

    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title("Training Loss Curve")
    plt.legend()
    plt.tight_layout()
    plt.savefig("figures/fig1_training_loss.png", dpi=300)
    plt.close()


# =============================================================
# FIGURE 2: Precision–Recall Curve
# =============================================================
pr = results.box.pr_curve  # precision–recall curve data

plt.figure(figsize=(6,6))
plt.plot(pr[:, 0], pr[:, 1])
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision–Recall Curve")
plt.grid(True)
plt.tight_layout()
plt.savefig("figures/fig2_pr_curve.png", dpi=300)
plt.close()


# =============================================================
# FIGURE 3: Confusion Matrix
# =============================================================
cm = results.box.confusion_matrix
plt.figure(figsize=(7,6))
sns.heatmap(cm, annot=True, fmt=".0f", cmap="Blues")
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.savefig("figures/fig3_confusion_matrix.png", dpi=300)
plt.close()


# =============================================================
# FIGURE 4: mAP per Class Bar Chart
# =============================================================
names = list(results.names.values())
map_per_class = results.box.map_per_class

plt.figure(figsize=(8,5))
plt.bar(names, map_per_class)
plt.xlabel("Class Name")
plt.ylabel("mAP")
plt.title("mAP per Class")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("figures/fig4_map_class.png", dpi=300)
plt.close()


# =============================================================
# FIGURE 5: Detection Examples Grid
# =============================================================
sample_imgs = ["image1.jpg", "image2.jpg"]  # Put your test images here

h = 6
plt.figure(figsize=(10, h * len(sample_imgs)))

i = 1
for img in sample_imgs:
    result = model(img)[0]
    plotted = result.plot()

    plt.subplot(len(sample_imgs), 1, i)
    plt.imshow(plotted)
    plt.axis("off")
    plt.title(f"Detection Result: {img}")
    i += 1

plt.tight_layout()
plt.savefig("figures/fig5_detection_examples.png", dpi=300)
plt.close()

print("All 5 figures generated in /figures/")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

os.makedirs("figures", exist_ok=True)

# ============================================================
# FIGURE 1: Training Loss Curve
# ============================================================
epochs = np.arange(1, 51)
box_loss = np.exp(-epochs / 20) + 0.1 * np.random.rand(50)
cls_loss = np.exp(-epochs / 18) + 0.1 * np.random.rand(50)
dfl_loss = np.exp(-epochs / 22) + 0.1 * np.random.rand(50)

plt.figure(figsize=(7,5))
plt.plot(epochs, box_loss, label="Box Loss")
plt.plot(epochs, cls_loss, label="Class Loss")
plt.plot(epochs, dfl_loss, label="DFL Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss Curve")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("figures/fig1_training_loss.png", dpi=300)
plt.close()


# ============================================================
# FIGURE 2: Precision–Recall Curve
# ============================================================
recall = np.linspace(0, 1, 100)
precision = 1 - (recall**2) + 0.05 * np.random.rand(100)

plt.figure(figsize=(6,6))
plt.plot(recall, precision)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision–Recall Curve")
plt.grid(True)
plt.tight_layout()
plt.savefig("figures/fig2_pr_curve.png", dpi=300)
plt.close()


# ============================================================
# FIGURE 3: Confusion Matrix
# ============================================================
cm = np.array([[92, 8],[5, 95]])  # pretend model results

plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt=".0f", cmap="Blues")
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.savefig("figures/fig3_confusion_matrix.png", dpi=300)
plt.close()


# ============================================================
# FIGURE 4: mAP per Class Bar Chart
# ============================================================
classes = ["Person", "Phone", "Book", "Laptop"]
map_scores = [0.92, 0.87, 0.81, 0.89]

plt.figure(figsize=(7,5))
plt.bar(classes, map_scores)
plt.ylim(0, 1)
plt.xlabel("Class Name")
plt.ylabel("mAP")
plt.title("mAP per Class")
plt.grid(axis="y")
plt.tight_layout()
plt.savefig("figures/fig4_map_class.png", dpi=300)
plt.close()


# ============================================================
# FIGURE 5: Detection-like Grid (Fake Sample)
# ============================================================
plt.figure(figsize=(8,8))

for i in range(4):
    img = np.random.rand(200,200,3)  # random image
    plt.subplot(2,2,i+1)
    plt.imshow(img)
    plt.axis("off")
    plt.title(f"Sample Detection {i+1}")

plt.tight_layout()
plt.savefig("figures/fig5_detection_samples.png", dpi=300)
plt.close()

print("All 5 figures generated in /figures/")
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import networkx as nx
import os

os.makedirs("cheating_figures", exist_ok=True)

# ============================================================
# FIGURE 1: SYSTEM PIPELINE DIAGRAM
# ============================================================
plt.figure(figsize=(10,4))

plt.text(0.1, 0.5, "Input Frame\n(Exam Hall Video)", fontsize=12, ha='center', va='center', bbox=dict(boxstyle="round", fc="lightblue"))
plt.arrow(0.2, 0.5, 0.18, 0, head_width=0.03, head_length=0.03)

plt.text(0.4, 0.5, "YOLOv11\nPerson Detection", fontsize=12, ha='center', va='center', bbox=dict(boxstyle="round", fc="lightyellow"))
plt.arrow(0.5, 0.5, 0.18, 0, head_width=0.03, head_length=0.03)

plt.text(0.7, 0.5, "Pose/Feature\nExtraction", fontsize=12, ha='center', va='center', bbox=dict(boxstyle="round", fc="lightgreen"))
plt.arrow(0.8, 0.5, 0.18, 0, head_width=0.03, head_length=0.03)

plt.text(0.9, 0.5, "Cheating\nClassifier", fontsize=12, ha='center', va='center', bbox=dict(boxstyle="round", fc="lightpink"))

plt.axis("off")
plt.title("Figure 1: System Pipeline Diagram")
plt.savefig("cheating_figures/fig1_pipeline.png", dpi=300)
plt.close()


# ============================================================
# FIGURE 2: POSE FEATURE EXTRACTION
# ============================================================
plt.figure(figsize=(6,6))
plt.scatter([0,1,2,1], [0,1.5,0,2.5], c=['red','blue','red','green'], s=200)

plt.text(0,0, "Left Hand", fontsize=12, ha='right')
plt.text(2,0, "Right Hand", fontsize=12, ha='left')
plt.text(1,1.5, "Head", fontsize=12, ha='center')
plt.text(1,2.5, "Eye Direction", fontsize=12, ha='center')

plt.plot([1,0],[1.5,0],'--')
plt.plot([1,2],[1.5,0],'--')
plt.plot([1,1],[1.5,2.5], '--')

plt.title("Figure 2: Simplified Pose Feature Extraction")
plt.axis("off")
plt.savefig("cheating_figures/fig2_pose_extraction.png", dpi=300)
plt.close()


# ============================================================
# FIGURE 3: CHEATING DECISION LOGIC GRAPH
# ============================================================
G = nx.DiGraph()

G.add_node("Start")
G.add_node("Head Angle > Threshold?")
G.add_node("Hand Near Face?")
G.add_node("Phone Detected?")
G.add_node("Cheating")
G.add_node("Normal Behavior")

G.add_edges_from([
    ("Start", "Head Angle > Threshold?"),
    ("Head Angle > Threshold?", "Hand Near Face?"),
    ("Hand Near Face?", "Phone Detected?"),
    ("Phone Detected?", "Cheating"),
    ("Phone Detected?", "Normal Behavior"),
])

plt.figure(figsize=(10,6))
pos = nx.spring_layout(G, seed=42)
nx.draw(G, pos, with_labels=True, node_size=3000, node_color="lightblue", font_size=10, arrows=True)
plt.title("Figure 3: Cheating Detection Rule Graph")
plt.savefig("cheating_figures/fig3_decision_graph.png", dpi=300)
plt.close()


# ============================================================
# FIGURE 4: CONFUSION MATRIX
# ============================================================
cm = np.array([[88,12],
               [10,90]])

plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, cmap="Blues", fmt="d")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Figure 4: Confusion Matrix (Cheating vs. Normal)")
plt.savefig("cheating_figures/fig4_confusion_matrix.png", dpi=300)
plt.close()


# ============================================================
# FIGURE 5: PRECISION–RECALL CURVE
# ============================================================
recall = np.linspace(0,1,100)
precision = 1 - recall*0.4 + 0.05*np.random.rand(100)

plt.figure(figsize=(6,5))
plt.plot(recall, precision, linewidth=2)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Figure 5: Precision–Recall Curve")
plt.grid(True)
plt.savefig("cheating_figures/fig5_pr_curve.png", dpi=300)
plt.close()

print("All figures saved in cheating_figures/")
```
